## Setup and Configuration

```python
import pickle
import torch
import pandas as pd
from model import load_model
from inference import predict_video, predict_from_csv, predict_live_webcam, batch_predict_videos
from utils import get_device
```

```python
# Configuration - UPDATE THESE PATHS
MODEL_PATH = 'Saved_10/gcn_model.pth'
ENCODER_PATH = 'Saved_10/label_encoder.pkl'

# Model parameters (should match training)
IN_CHANNELS = 3
HIDDEN_CHANNELS = 8
NUM_LAYERS = 3
NUM_HEADS = 2
NUM_CLASSES = 10
```

## Load Model and Label Encoder

```python
print("Loading model...")
device = get_device()

model = load_model(
    MODEL_PATH,
    in_channels=IN_CHANNELS,
    hidden_channels=HIDDEN_CHANNELS,
    out_channels=NUM_CLASSES,
    num_layers=NUM_LAYERS,
    num_heads=NUM_HEADS,
    device=device
)

print(f"Loading label encoder from {ENCODER_PATH}...")
with open(ENCODER_PATH, 'rb') as f:
    label_encoder = pickle.load(f)

model = model.to(device)
model.eval()

print(f"\n✓ Model loaded successfully on {device}")
print(f"✓ Number of classes: {len(label_encoder.classes_)}")
print(f"✓ Classes: {list(label_encoder.classes_)}")
```

## Model Information

```python
print("Model Architecture:")
print(model)

total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"\nTotal parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
```

## Single Video Prediction

```python
video_path = r"KArSL\combined_test\0001-0070\3\01_01_0003_(10_11_16_16_28_15)_c.mp4"

print(f"Predicting video: {video_path}")
predicted_label = predict_video(video_path, model, label_encoder, device=device, T=48)

print(f"\n🎯 Predicted Sign: {predicted_label}")
```
